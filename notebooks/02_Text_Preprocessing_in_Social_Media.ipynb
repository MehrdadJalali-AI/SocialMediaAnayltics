## Concepts in simple language

Lowercasing, whitespace, URLs, mentions, hashtags, emojis, punctuation, numbers, contractions, repeated letters, tokenization, stop words, stemming and lemmatization change the signal. Negation and emoji often carry sentiment; sarcasm and multilingual text remain hard. Excessive cleaning can erase meaning.

## Instructor-guided walkthrough

Each function validates input, has a type hint and docstring, and performs one visible transformation.

## Step-by-step Python implementation

```python
import re
def _check(text: str) -> str:
    """Validate and return text."""
    if not isinstance(text,str): raise TypeError('text must be a string')
    return text
def normalize_whitespace(text: str) -> str:
    """Collapse repeated whitespace. Example: 'a  b' -> 'a b'."""
    return re.sub(r'\s+',' ',_check(text)).strip()
def replace_urls(text: str) -> str:
    """Replace web links with the token URL."""
    return re.sub(r'https?://\S+','URL',_check(text))
def replace_mentions(text: str) -> str:
    """Replace @handles with USER to reduce identifiability."""
    return re.sub(r'@\w+','USER',_check(text))
def split_hashtag(text: str) -> str:
    """Remove # but preserve the hashtag word."""
    return re.sub(r'#(\w+)',r'\1',_check(text))
def reduce_repeated_characters(text: str) -> str:
    """Reduce 3+ repeated characters to two."""
    return re.sub(r'(.)\1{2,}',r'\1\1',_check(text))
def preprocess_social_text(text: str) -> str:
    """Apply the teaching pipeline; preserve emoji and words such as not."""
    value=normalize_whitespace(text.lower())
    for fn in (replace_urls,replace_mentions,split_hashtag,reduce_repeated_characters): value=fn(value)
    return value
```

```python
examples=posts=pd.read_csv(data_directory/'social_media_sample.csv').dropna(subset=['text']).head(8).copy()
examples['cleaned_text']=examples['text'].map(preprocess_social_text)
examples['tokens']=examples['cleaned_text'].str.findall(r'\b\w+\b|[^\w\s]')
examples['token_count']=examples['tokens'].str.len()
display(examples[['text','cleaned_text','tokens','token_count']])
print('Before/after:',examples.iloc[0]['text'],'=>',examples.iloc[0]['cleaned_text'])
```

The comparison makes every loss visible. `not`, hashtag words, and emoji remain because they can alter sentiment. Tokenization separates units; stemming cuts word endings, while lemmatization seeks dictionary forms.
