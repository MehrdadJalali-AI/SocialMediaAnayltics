## Concepts in simple language

Social media analytics studies text, metadata, engagement and networks. Sentiment is polarity (positive, neutral, negative). Document-, sentence-, and aspect-level analysis differ in granularity. Emotion names feelings; stance expresses support or opposition. Sources and sampling create platform, self-selection, temporal, and model bias. GDPR, privacy, consent, and platform rules constrain responsible work.

## Dataset introduction

`social_media_sample.csv` contains posts, labels, engagement, time, and topic fields. All data are synthetic teaching data; they contain no real private users.

```python
from pathlib import Path
import random, sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
RANDOM_SEED = 42
random.seed(RANDOM_SEED); np.random.seed(RANDOM_SEED)
current_path = Path.cwd()
if (current_path / 'data').exists(): repository_root = current_path
elif (current_path.parent / 'data').exists(): repository_root = current_path.parent
else: raise FileNotFoundError('Could not find the repository data directory.')
data_directory = repository_root / 'data'
print('Python:', sys.version.split()[0], '| pandas:', pd.__version__)
print('Data directory:', data_directory)
```

## Instructor-guided walkthrough

We load the CSV, inspect its shape, schema, missing values, duplicate IDs, text lengths, time coverage, language and sentiment balance.

```python
posts=pd.read_csv(data_directory/'social_media_sample.csv')
posts['created_at']=pd.to_datetime(posts['created_at'])
print('Shape:',posts.shape); display(posts.head(3)); display(posts.dtypes.to_frame('dtype'))
```

```python
display(posts.isna().sum().to_frame('missing'))
print('Duplicate full rows:',posts.duplicated().sum(),'Duplicate post IDs:',posts['post_id'].duplicated().sum())
posts['text_length']=posts['text'].fillna('').str.len()
print('Time range:',posts['created_at'].min(),'to',posts['created_at'].max())
display(posts['language'].value_counts(dropna=False).to_frame('posts'))
```

```python
counts=posts['sentiment'].value_counts(); percentages=posts['sentiment'].value_counts(normalize=True).mul(100).round(1)
engagement=posts.assign(total_engagement=posts[['likes','shares','replies']].sum(axis=1)).groupby('sentiment')['total_engagement'].mean().round(1)
display(pd.concat([counts.rename('count'),percentages.rename('percent')],axis=1)); display(engagement.to_frame())
counts.plot(kind='bar',title='Synthetic posts by sentiment'); plt.xlabel('Sentiment'); plt.ylabel('Number of posts'); plt.tight_layout(); plt.show()
```

The class counts are deliberately close, while missing and duplicated records make the quality audit realistic. The chart describes this sample only.
